```python
import pandas as pd

excel_path = ("../data/raw/classification/Book1.xlsx")

#inspect the namnes without reading all data into memory
excel_file = pd.ExcelFile(excel_path)
print("Found Sheets:", excel_file.sheet_names)
```

```python
header_rows = {
    "COMMUNICATION":2,
    "CLEARANCE": 1,
    "AOsMEMOs": 3,
    "LEAVES": 0,
}

for sheets in excel_file.sheet_names:
    header_row = header_rows.get(sheets,0)
    
    df = pd.read_excel(
        excel_file,
        sheet_name=sheets,
        header=header_row,

    )
    
    print(f"\n--- [{sheets}] ---")
    display(df)
```

```python
print(df.shape) #dimensionality: (rows, columns)
print(df.columns.tolist()) #schema / feature names
print(df.dtypes) #data types inferred by pandas
print(df.isna().sum()) #missing-value detection
print(df.duplicated().sum()) #check duplicate records
```

```python
print(f"\n---{sheets}---")
print("Shape", df.shape)
print("Duplicates:", df.duplicated().sum())
print("\nMissing values:")
print(df.isna().sum())
```

```python
#investigate the 3 duplicates 
duplicates = df[df.duplicated(keep=False)]
display(duplicates)
```

```python
#detect empty rows
empty_rows  = df[df.isna().all(axis=1)]
display(empty_rows)
```

```python
#how="all" means : drop a row only when every column is Null
df = df.dropna(how="all")
```

```python
#inspect categorical columns
print(df["TYPE OF LEAVE"].value_counts(dropna=False))
print(df["STATUS"].value_counts(dropna=False))
```

```python
df["TYPE OF LEAVE"] = (
    df["TYPE OF LEAVE"]
    .str.strip()
    .str.upper()
)

#You should see some categories merge.
df["TYPE OF LEAVE"].value_counts(dropna=False)

```
